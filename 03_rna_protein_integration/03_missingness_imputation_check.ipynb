Script to check number of imputation per classification category per cell type

```python
import pandas as pd
```

```python
path = "results"
xlsx_names = ["ver_10_jackson_goblet_concordance_full.xlsx", "ver_10_jackson_club_concordance_full.xlsx", "ver_10_jackson_ciliated_concordance_full.xlsx"]
```

```python
# --- goblet (max 4 replicates per condition) ---
sheet = pd.read_excel(path + xlsx_names[0], sheet_name="Protein_only")
cls   = "protein_class_jackson_goblet"
MAX   = 4

d = sheet[[cls]].dropna().copy()
d[['obs_ctrl','obs_il13']] = d[cls].astype(str).str.split('-', expand=True).astype(int)

def imputation_check(oc, oi, mx=MAX):
    if oc == mx and oi == mx:                       return "Fully observed"
    if oc >= 2 and oi >= 2:                          return "Limited missingness"
    if (oc == mx and oi <= 1) or (oi == mx and oc <= 1): return "One-sided"
    return "Sparse / heavily imputed"

d['missing_class'] = [imputation_check(a, b, MAX) for a, b in zip(d['obs_ctrl'], d['obs_il13'])]

n = len(d)
tab = d['missing_class'].value_counts()
print("Total protein-only:", n)
print((100 * tab / n).round(1))
```

```python
# --- club (max 4 replicates per condition) ---
sheet = pd.read_excel(path + xlsx_names[1], sheet_name="Protein_only")
cls   = "protein_class_jackson_club"
MAX   = 4

d = sheet[[cls]].dropna().copy()
d[['obs_ctrl','obs_il13']] = d[cls].astype(str).str.split('-', expand=True).astype(int)

d['missing_class'] = [imputation_check(a, b, MAX) for a, b in zip(d['obs_ctrl'], d['obs_il13'])]

n = len(d)
tab = d['missing_class'].value_counts()
print("Total protein-only:", n)
print((100 * tab / n).round(1))
```

```python
# --- ciliated (max 3 replicates per condition) ---
sheet = pd.read_excel(path + xlsx_names[2], sheet_name="Protein_only")
cls   = "protein_class_jackson_ciliated"
MAX   = 3

d = sheet[[cls]].dropna().copy()
d[['obs_ctrl','obs_il13']] = d[cls].astype(str).str.split('-', expand=True).astype(int)

d['missing_class'] = [imputation_check(a, b, MAX) for a, b in zip(d['obs_ctrl'], d['obs_il13'])]

n = len(d)
tab = d['missing_class'].value_counts()
print("Total protein-only:", n)
print((100 * tab / n).round(1))
```
